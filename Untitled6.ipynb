```python
# ============================================
# SCI RECOVERY PREDICTION — FINAL DEMO MODEL
# ============================================

import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression

# -----------------------------
# 1. Load data
# -----------------------------

df = pd.read_excel("SCI_ML_Demo_Data_v2.xlsx")

# -----------------------------
# 2. Define outcome
# -----------------------------

ais_order = {
    "A": 0,
    "B": 1,
    "C": 2,
    "D": 3,
    "E": 4
}

baseline = df["Initial_AIS"].map(ais_order)
followup = df["AIS_6M"].map(ais_order)

df["Recovery"] = (
    (followup - baseline) >= 1
).astype(int)

# -----------------------------
# 3. Define predictors
# -----------------------------

features = [
    "Age",
    "Sex",
    "Injury_Level",
    "Initial_AIS",
    "Initial_Motor",
    "Edema",
    "Edema_Length_mm",
    "Hemorrhage",
    "Hemorrhage_Length_mm",
    "Canal_Compromise_percent"
]

X = df[features]
y = df["Recovery"]

# -----------------------------
# 4. Identify variable types
# -----------------------------

categorical_features = [
    "Sex",
    "Injury_Level",
    "Initial_AIS",
    "Edema",
    "Hemorrhage"
]

numeric_features = [
    "Age",
    "Initial_Motor",
    "Edema_Length_mm",
    "Hemorrhage_Length_mm",
    "Canal_Compromise_percent"
]

# -----------------------------
# 5. Preprocessing
# -----------------------------

preprocessor = ColumnTransformer([
    (
        "num",
        SimpleImputer(strategy="median"),
        numeric_features
    ),
    (
        "cat",
        Pipeline([
            ("imputer",
             SimpleImputer(strategy="most_frequent")),
            ("onehot",
             OneHotEncoder(handle_unknown="ignore"))
        ]),
        categorical_features
    )
])

# -----------------------------
# 6. Machine-learning model
# -----------------------------

final_model = Pipeline([
    ("preprocessing", preprocessor),
    ("model", LogisticRegression(max_iter=1000))
])

# -----------------------------
# 7. Train
# -----------------------------

final_model.fit(X, y)

print("FINAL MODEL TRAINED SUCCESSFULLY")
print("Patients:", len(df))
print("Predictors:", len(features))
```

```python
import numpy as np
from sklearn.metrics import roc_auc_score

y_array = np.asarray(y)
prob_array = np.asarray(mri_oof_probability)

rng = np.random.default_rng(42)

bootstrap_auc = []

for i in range(2000):

    indices = rng.integers(
        0,
        len(y_array),
        size=len(y_array)
    )

    y_boot = y_array[indices]
    prob_boot = prob_array[indices]

    if len(np.unique(y_boot)) < 2:
        continue

    bootstrap_auc.append(
        roc_auc_score(y_boot, prob_boot)
    )

bootstrap_auc = np.array(bootstrap_auc)

lower = np.percentile(bootstrap_auc, 2.5)
upper = np.percentile(bootstrap_auc, 97.5)

print("Clinical + MRI AUC:", round(mri_auc, 3))
print("95% CI:", round(lower, 3), "-", round(upper, 3))
```

```python
# ==========================================
# VERSION 2 — MODEL EXPLANATION
# ==========================================

def explain_prediction_v2(button):

    with output_v2:

        clear_output()

        patient_v2 = pd.DataFrame({
            "Age": [age_v2.value],
            "Sex": [sex_v2.value],
            "Injury_Level": [level_v2.value],
            "Initial_AIS": [ais_v2.value],
            "Initial_Motor": [motor_v2.value],
            "Edema": [edema_v2.value],
            "Edema_Length_mm": [edema_length_v2.value],
            "Hemorrhage": [hemorrhage_v2.value],
            "Hemorrhage_Length_mm": [hemorrhage_length_v2.value],
            "Canal_Compromise_percent": [canal_v2.value]
        })

        probability = mri_model.predict_proba(patient_v2)[0, 1]

        # Extract preprocessing and logistic regression model
        preprocessor = mri_model.named_steps["preprocessing"]
        model = mri_model.named_steps["model"]

        # Transform patient data
        transformed = preprocessor.transform(patient_v2)

        # Get feature names
        feature_names = preprocessor.get_feature_names_out()

        # Logistic regression contribution
        coefficients = model.coef_[0]

        # Contribution to log-odds
        contributions = transformed[0] * coefficients

        # Convert sparse matrix if necessary
        if hasattr(contributions, "toarray"):
            contributions = contributions.toarray().ravel()
        else:
            contributions = np.asarray(contributions).ravel()

        # Top contributors
        contribution_df = pd.DataFrame({
            "Feature": feature_names,
            "Contribution": contributions
        })

        contribution_df["Absolute"] = contribution_df["Contribution"].abs()

        contribution_df = (
            contribution_df
            .sort_values("Absolute", ascending=False)
            .head(6)
        )

        display(HTML(f"""
        <div style="
            background:#f5f7fa;
            padding:20px;
            border-radius:12px;
            border:1px solid #ddd;">

            <h2 style="text-align:center;">
                Why did the model make this prediction?
            </h2>

            <div style="text-align:center;
                        font-size:40px;
                        font-weight:bold;
                        margin:15px;">
                {probability*100:.1f}%
            </div>

            <p style="text-align:center;">
                Predicted probability of ≥1 AIS-grade improvement
                at 6 months
            </p>

            <hr>

            <h3>Important model contributors</h3>

        """))

        for _, row in contribution_df.iterrows():

            feature = row["Feature"]
            contribution = row["Contribution"]

            if contribution > 0:
                direction = "↑ contributes toward prediction"
            else:
                direction = "↓ contributes away from prediction"

            print(
                f"{feature}: "
                f"{contribution:.3f}   {direction}"
            )

        display(HTML("""
            <hr>
            <p style="
                text-align:center;
                color:#777;
                font-size:12px;">
                Research prototype using synthetic demonstration data.
                Not for clinical decision-making.
            </p>
        </div>
        """))

# Replace previous button callback
predict_v2._click_handlers.callbacks = []
predict_v2.on_click(explain_prediction_v2)

print("✅ Prediction + explanation engine ready.")
```
